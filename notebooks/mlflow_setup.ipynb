# MLflow + MinIO connection setup

Verify the local experiment tracking stack is running.

**Prerequisites:** `cd mlflow && docker compose up -d` (wait ~30s for all services to be healthy)

```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path('.').resolve()))

from chronos_ts.tracking import configure_mlflow

exp_id = configure_mlflow(
    tracking_uri='http://localhost:5050',
    experiment_name='chronos-1h-classification',
    s3_endpoint_url='http://localhost:9000',
)
```

```python
import mlflow

# Verify: list experiments
exps = mlflow.search_experiments()
for e in exps:
    print(f"  id={e.experiment_id:3s}  name={e.name}  artifact_location={e.artifact_location}")
```

```python
import boto3
import os

# Verify: list MinIO buckets
s3 = boto3.client(
    's3',
    endpoint_url=os.environ['MLFLOW_S3_ENDPOINT_URL'],
    aws_access_key_id=os.environ['AWS_ACCESS_KEY_ID'],
    aws_secret_access_key=os.environ['AWS_SECRET_ACCESS_KEY'],
)
response = s3.list_buckets()
print('Buckets:', [b['Name'] for b in response.get('Buckets', [])])
```
